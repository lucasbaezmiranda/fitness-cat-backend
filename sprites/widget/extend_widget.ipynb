```python
from PIL import Image
import os

def extend_image_background_right(input_path, output_suffix="_extended"):
    try:
        # 1. Cargar la imagen original
        img = Image.open(input_path)
        orig_width, orig_height = img.size
        img_mode = img.mode

        # 2. Calcular nuevas dimensiones (duplicar ancho)
        new_width = orig_width * 2
        new_height = orig_height

        # 3. Crear un nuevo lienzo en blanco del doble de ancho
        new_img = Image.new(img_mode, (new_width, new_height))

        # 4. Pegar la imagen original en la parte izquierda del nuevo lienzo
        new_img.paste(img, (0, 0))

        # 5. Método de relleno: repetir la columna de píxeles del borde derecho
        # Extraemos una tira de 1 píxel de ancho de todo el alto del borde derecho
        rightmost_pixel_strip = img.crop((orig_width - 1, 0, orig_width, orig_height))

        # Rellenamos el nuevo espacio a la derecha pegando esta tira repetidamente
        for x in range(orig_width, new_width):
            new_img.paste(rightmost_pixel_strip, (x, 0))

        # 6. Construir el nombre del archivo de salida
        file_root, file_ext = os.path.splitext(input_path)
        output_path = f"{file_root}{output_suffix}{file_ext}"

        # 7. Guardar la nueva imagen extendida
        new_img.save(output_path)
        print(f"Imagen extendida guardada exitosamente en: {output_path}")

        # Devolvemos el objeto imagen para que Jupyter lo renderice automáticamente
        return new_img

    except FileNotFoundError:
        print(f"Error: No se pudo encontrar el archivo '{input_path}'. Verifica que esté en la misma carpeta que el notebook.")
        return None
    except Exception as e:
        print(f"Ocurrió un error inesperado: {e}")
        return None

# --- CONFIGURACIÓN Y EJECUCIÓN ---

# Define aquí el nombre de tu archivo de imagen
# (Debe estar en la misma carpeta donde se está ejecutando este notebook)
ruta_imagen = "widget_mad.jpeg"

# Ejecutamos la función
imagen_final = extend_image_background_right(ruta_imagen)

# Al poner la variable sola al final de la celda, 
# Jupyter Notebook mostrará la imagen visualmente debajo del código.
imagen_final
```
